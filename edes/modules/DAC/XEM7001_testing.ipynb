```python
import matplotlib.pyplot as plt
import numpy as np
from FPGA import *
from edes.utils.plotting import plot, plot_ax, big_plt_font 
big_plt_font()

board = FPGA()
board.restart()
fig, ax = plt.subplots(ncols=4, nrows=2, figsize=(40, 20))
fig.tight_layout()
for x in range(8):
    voltage = []
    binary = []
    back = []
    for y in range(10):
        board.all_off()
        voltage.append(1.8/10 * y)
        s = board.set_voltage(x,1.8/10 *y)
        #print(x,y)
        #print(int("".join(map(str, s)),2))
        binary.append(int("".join(map(str, s)),2))
        board.send()
        ba = board.read()
        #print(ba)
        back.append(int("".join(ba),2))
    # print("v",voltage)
    print("b", binary)
    #print("ba", back)
    plot_ax(ax[x//4][x%4], voltage, binary, '.--', label="binary in", xlabel='xxx', ylabel='yyy')
    plot_ax(ax[x//4][x%4], voltage, back, '.--', label="binary out", xlabel=f'Channel {x+1} voltage (V)', ylabel='digital code (decimal)')
plt.legend() 
plt.tight_layout() 
plt.show()
board.board.dev.Close()
```
