```python
class AnharmonicElectron(HarmonicElectron): 
    def __init__(self, n, d, omega, Qe, Te=4, m=m, e=e, name='Hot_electron'): 
        super().__init__(n, d, omega, m=m, e=e, name=name) 
        self.r = np.sqrt(self.l/self.c) / Qe 
        self.Qe = Qe 
        self.Te = Te 

    def get_noise(self, f): 
        PSD = 4*kB*self.Te*self.r
        return self.white_noise(f, PSD)

    def get_rlc(self): 
        return self.r, self.l, self.c

class TankCircuitPinMatchedAmp(TankCircuit): 
    def __init__(self, Q, Z0, omega_0, Cc, ZL, Tr=4, Ta=4, name='TankCircuit'): 
        super().__init__(Q, Z0, omega_0, T=Tr) 
        self.T = Tr 
        self.Tr = Tr 
        self.Ta = Ta 
        self.Cc = Cc 
        self.Zload = ZL 

    def get_noise_tank(self, f): 
        return super().get_noise(f) 

    def get_noise_amp(self, f): 
        PSD = 4*kB*self.Ta*self.Zload
        return self.white_noise(f, PSD)

class TankCircuitAbstractMatched(TankCircuit): 
    def __init__(self, Q, Z0, omega_0, match_loss, ZL, Tr=4, Ta=4, name='TankCircuit'): 
        super().__init__(Q, Z0, omega_0, T=Tr) 
        self.T = Tr 
        self.Tr = Tr 
        self.Ta = Ta 
        self.match_loss = match_loss 
        self.Zload = ZL 

    def get_noise_tank(self, f): 
        return super().get_noise(f)

    def get_noise_amp(self, f): 
        PSD = 4*kB*self.Ta*self.Zload
        return self.white_noise(f, PSD)
```

```python
class AnharmonicElectronCoupledTankWithAmp(Component): 
    def __init__(self, n, d, Qr, Qe, Z0, omega_0, omega_z, Cc,
                 m=m, e=e, Tr=4, Te=4, Ta=4, ZL=50,
                 name='Electron'):
        super().__init__(name=name, n=n, d=d, Qe=Qe, Qr=Qr, Z0=Z0,
                         omega_0=omega_0, omega_z=omega_z, Cc=Cc,
                         Tr=Tr, Te=Te, Ta=Ta, ZL=ZL, m=m, e=e)
        self.Electron = AnharmonicElectron(n, d, omega_z, Qe, Te, m, e, name)
        self.TankCircuit = TankCircuitPinMatchedAmp(Qr, Z0, omega_0, Cc, ZL, 
                                                    Tr, Ta, name)
        self.R, self.L, self.C = self.TankCircuit.get_RLC()
        self.r, self.l, self.c = self.Electron.get_rlc()
        
    def Hr(self, omega):
        '''
        series rlc (electron) coupled to a parallel RLC (tank circuit) 
        circuit, see Wineland 1975
        '''
        s = 1j*omega
        s = 1j*omega
        Zl = s*self.l
        ZL = s*self.L
        Zc = 1/(s*self.c)
        ZC = 1/(s*self.C)
        ZCc = 1/(s*self.Cc)
        R = self.R
        r = self.r
        Rload = self.Rload
        
        Zlc = parallel( (Zl+Zc+r), parallel(ZL, ZC) )
        Zin_1 = parallel( Zlc, (ZCc+Rload) )
        return (Zin_1 / (Zin_1+R)) * (Rload / (Rload+ZCc))

    def He(self, omega): 
        s = 1j*omega
        s = 1j*omega
        Zl = s*self.l
        ZL = s*self.L
        Zc = 1/(s*self.c)
        ZC = 1/(s*self.C)
        ZCc = 1/(s*self.Cc)
        R = self.R
        r = self.r
        Rload = self.Rload
        
        Z_tank = parallel( R, parallel(ZL, ZC) )
        Zin_1 = parallel( Z_tank, (ZCc+Rload) )
        Z_ele = r + Zl + Zc
        return (Zin_1 / (Zin_1+Z_ele)) * (Rload / (Rload+ZCc))

    def __call__(self, freq): 
        tank_noise = self.TankCircuit.get_noise_tank(freq)
        amp_noise = self.TankCircuit.get_noise_amp(freq)
        ele_noise = self.Electron.get_noise(freq)
        omega = 2*np.pi*freq 
        return (abs(self.He(omega))**2 * abs(ele_noise) + \
               abs(self.Hr(omega))**2 * abs(tank_noise)) +  \
               abs(amp_noise)
```

```python
class AnharmonicElectronCoupledTankWithMatch(Component): 
    def __init__(self, n, d, Qr, Qe, Z0, omega_0, omega_z, matching_loss,
                 m=m, e=e, Tr=4, Te=4, Ta=4, ZL=50,
                 name='Electron'):
        super().__init__(name=name, n=n, d=d, Qe=Qe, Qr=Qr, Z0=Z0,
                         omega_0=omega_0, omega_z=omega_z, matching_loss=matching_loss,
                         Tr=Tr, Te=Te, Ta=Ta, ZL=ZL, m=m, e=e)
        self.Electron = AnharmonicElectron(n, d, omega_z, Qe, Te, m, e, name)
        self.TankCircuit = TankCircuitAbstractMatched(Qr, Z0, omega_0, matching_loss, ZL, 
                                                    Tr, Ta, name)
        self.R, self.L, self.C = self.TankCircuit.get_RLC()
        self.r, self.l, self.c = self.Electron.get_rlc()
        
    def Hr(self, omega):
        '''
        series rlc (electron) coupled to a parallel RLC (tank circuit) 
        circuit, see Wineland 1975
        '''
        s = 1j*omega
        s = 1j*omega
        Zl = s*self.l
        ZL = s*self.L
        Zc = 1/(s*self.c)
        ZC = 1/(s*self.C)
        R = self.R
        r = self.r
        
        Zlc = parallel( (Zl+Zc+r), parallel(ZL, ZC) )
        return (Zlc / (Zlc+R))
    

    def He(self, omega): 
        s = 1j*omega
        s = 1j*omega
        Zl = s*self.l
        ZL = s*self.L
        Zc = 1/(s*self.c)
        ZC = 1/(s*self.C)
        R = self.R
        r = self.r
        
        Z_tank = parallel( R, parallel(ZL, ZC) )
        Z_ele = r + Zl + Zc
        return (Z_tank / (Z_tank+Z_ele))

    def __call__(self, freq): 
        tank_noise = self.TankCircuit.get_noise_tank(freq)
        amp_noise = self.TankCircuit.get_noise_amp(freq)
        ele_noise = self.Electron.get_noise(freq)
        omega = 2*np.pi*freq 
        return (abs(self.He(omega))**2 * abs(ele_noise) + \
               abs(self.Hr(omega))**2 * abs(tank_noise)) / self.matching_loss +  \
               abs(amp_noise)
```

```python
class HotElectronSetup(DetectionSetup):
    def __init__(self, n, d, Qr, Qe, matching_loss, Z0, f0, fz_0, 
                 Te, Tr, Ta,
                 BW_measurement, Z_measurement, noise_floor, 
                 N_average, m=m, e=e,
                 devices=[],
                 display_progress=False, 
		         display_averaging=False,
                 name='AnharmonicElectron'):
        
        super().__init__(BW_measurement, Z_measurement, noise_floor, 
                         name=name, n=n, d=d, Qr=Qr, Z0=Z0, f0=f0, 
                         fz_0=fz_0, Qe=Qe, Tr=Tr, Ta=Ta,
                         m=m, e=e, Te=Te, N_average=N_average,
                         display_progress=display_progress)
        self.devices = devices
        self.electron = AnharmonicElectronCoupledTankWithMatch(
                            n, d, Qr, Qe, Z0, 
                            2*np.pi*f0, 2*np.pi*fz_0, matching_loss,
                            m=m, e=e, Tr=Tr, Te=Te, Ta=Ta, ZL=Z_measurement)
        self.display_averaging = display_averaging

    def __call__(self, freq):
        spectrum = np.zeros(np.shape(freq))
        N_average = self.N_average
        
        for idx in trange(N_average, disable=(not self.display_averaging)):
            psd_signal = self.electron(freq)
                
            for device in self.devices:
                psd_signal = device(freq, psd_signal)
            
            ideal_ps, real_ps = self.measure_spectrum(freq, psd_signal)
            spectrum += dBm_to_W(real_ps) / N_average
        return W_to_dBm(spectrum)
```

```python
def find_SNR(x, y, debug):
    i_min = len(y)//2
    i_max1 = np.argmax(y[len(y)//5:i_min])+len(y)//5
    i_max2 = np.argmax(y[i_min:-len(y)//5])+i_min
    y_min = np.min([i for i in y[i_max1+1:i_max2] if not np.isnan(i)])  # adjust for nan
    i_y_min = [i for i in np.where(y==y_min) if i > i_max1 and i < i_max2][0]
    y_max = np.max(y[~np.isnan(y)])
    if debug:
        plot(x, y)
        plot(x[np.where(y==y_max)], y_max, 'X', markersize=10)
        plot(x[i_y_min], y_min, 'o', markersize=10)
    return y_max - y_min
    
def find_SNR_with_err(x, y, n=10, debug=False):
    """
    Find the SNR in dB and the associated errorbar.
    The error bars are found by finding the SNR again
    on a more denoised signal, by applying a moving average
    with size n*signal_resolution. NOTE: not used rn...
    """
    noisy_SNR = find_SNR(x, y, debug)
    smooth_SNR = find_SNR(x, moving_avg(y, window=n, display_progress=False), debug)
    return smooth_SNR, abs(noisy_SNR-smooth_SNR)

def plot_SNR_variations(all_width_factor, all_deff, all_SNR, Q, Z0, T, n, f0=70):
    fig, ax = plt.subplots(figsize=(8, 5))
    ax.set_xscale('log')
    ax2 = ax.twiny()
    ax2.set_xscale('log')
    for i in range(len(all_width_factor)):
        plot_ax(ax, all_deff*1e6, all_SNR[i], 'o--', label=f'$\Delta f/f_0 = {all_width_factor[i]:.0e}$')
        plot_ax(ax2, all_deff*1e6, all_SNR[i], '.')
    ax.set_ylabel('SNR (dB)')
    ax.set_xlabel('$d_{eff}$ (um)')
    ax2.set_xlabel('$\gamma/2\pi$ (KHz)')
    ax2.set_xticklabels([f'{i:.3f}' for i in e**2*Q*Z0/m/(ax2.get_xticks()*1e-6)**2/2e3/np.pi])
    ax.set_title(f'$T = ${T}K, $f_0 = ${f0}MHz, $n = {n}$')
    ax.legend(bbox_to_anchor=(1.01, 1))
    plt.tight_layout()
    plt.show()
```

```python
def thermal_distr_freq(dw, Tth):
    """
    Get the boltzmann distribution of frequency, following
        P(delta_omega) = exp{ T(delta_omega) / Tth },
    where T(delta_omega) is a fit done in anharmonicity 
    studies.
    
    Parameters
    ----------
    dw : [rad/s]
        Delta omega
    Tth : [K]
        Thermal temperature
    """
    dw = dw/1e6  # fitting is done in [2*pi * MHz]
    return np.exp(-(269.32*dw - 49.19*dw**2 + 19.1*dw**3)/Tth)
```

```python
dw = 2e6*np.pi*np.linspace(0, 1, 1000000)
for Tth in [4, 20, 100, 300]:
    P_dw = thermal_distr_freq(dw, Tth)
    f_plot = (f0+dw/2/np.pi)/1e6  # in MHz
    width = dw[np.argmin(abs(P_dw-0.5))]/2/np.pi/1e3
    plot(f_plot, P_dw, label=f'$T = {Tth}$K, width = {width:.2f}KHz \n$\Delta f/f_0 = {width/f0*1e3:.1e}$')
plt.xlabel('Frequency (MHz)')
plt.ylabel('Probability')
plt.legend()
plt.show()
```
